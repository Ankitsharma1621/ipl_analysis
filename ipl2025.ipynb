```python
batman["Performance Score"] =batman['Runs']+ (batman['50s'] * 10)+ (batman['100s'] * 25)+ (batman['6s'] *2)
```

```python
performance_bar=batman[["Performance Score","Player Name"]].head(10)
performance_bar
```

```python
sns.barplot(x='Performance Score',y='Player Name',data=performance_bar,palette='rainbow')
```
